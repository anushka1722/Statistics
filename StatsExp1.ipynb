```python
##Missing values
df.isnull().sum()
```

```python
#Invalid zero values 
columnstocheck = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for column in columnstocheck:
    print(column, ":", (df[column] == 0).sum())
```

```python
##Percentage of invalid values

for column in columnstocheck:
    zero_count = (df[column] == 0).sum()
    percentage = (zero_count / len(df)) * 100
    print(f"{column}: {zero_count} zeros ({percentage:.2f}%)")
```

```python
#Duplicate values
df.duplicated().sum()
```

```python
##Outlier Analysis

plt.figure(figsize=(15, 8))

sns.boxplot(data=df)

plt.xticks(rotation=45)
plt.title("Boxplot of Pima Indians Diabetes Dataset Variables")
plt.show()
```

```python
##Simple outlier check

Q1 = df["Glucose"].quantile(0.25)
Q3 = df["Glucose"].quantile(0.75)

IQR = Q3 - Q1

outliers = df[(df["Glucose"] < Q1 - 1.5*IQR) | 
              (df["Glucose"] > Q3 + 1.5*IQR)]

print("Number of Glucose outliers:", len(outliers))
```
