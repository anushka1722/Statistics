```python
# Identify potential outliers using IQR

Q1 = db[['Glucose', 'BMI', 'Age']].quantile(0.25)
Q3 = db[['Glucose', 'BMI', 'Age']].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("IQR:")
print(IQR)

print("\nLower Limit:")
print(lower_limit)

print("\nUpper Limit:")
print(upper_limit)
```

```python
# Count potential outliers

outliers = ((db[['Glucose', 'BMI', 'Age']] < lower_limit) |
            (db[['Glucose', 'BMI', 'Age']] > upper_limit))

outliers.sum()
```
