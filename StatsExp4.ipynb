```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
db = pd.read_csv("Diabetes.csv")
```

```python
db.columns
```

```python
#Sample mean of glucose
sample_mean = db["Glucose"].mean()

print("Sample Mean of Glucose:", sample_mean)
```

```python
#Sample Proportion
sample_proportion = db["Outcome"].mean()

print("Sample Proportion of Diabetes:", sample_proportion)
```

```python
#95% Confidence Interval
glucose = db["Glucose"].dropna()

mean_glucose = glucose.mean()
std_glucose = glucose.std()
n = len(glucose)

confidence_level = 0.95
alpha = 1 - confidence_level

t_critical = stats.t.ppf(1 - alpha/2, df=n-1)

margin_of_error = t_critical * (std_glucose / np.sqrt(n))

lower_limit = mean_glucose - margin_of_error
upper_limit = mean_glucose + margin_of_error

print("Sample Mean:", mean_glucose)
print("95% Confidence Interval:", (lower_limit, upper_limit))
```

```python
#One sample t test
glucose = db["Glucose"].dropna()

t_statistic, p_value = stats.ttest_1samp(glucose, 120)

print("t-statistic:", t_statistic)
print("p-value:", p_value)
```

```python
#95% Confidence Interval for Diabetes Proportion
x = db["Outcome"].sum()
n = len(db["Outcome"])

p_hat = x / n

z_critical = stats.norm.ppf(0.975)

margin_of_error = z_critical * np.sqrt((p_hat * (1 - p_hat)) / n)

lower_proportion = p_hat - margin_of_error
upper_proportion = p_hat + margin_of_error

print("Sample Proportion:", p_hat)
print("95% Confidence Interval for Proportion:",
      (lower_proportion, upper_proportion))
```

```python
print("STATISTICAL ESTIMATION AND HYPOTHESIS TESTING")
print("-" * 50)

print("1. Sample Mean of Glucose:", round(sample_mean, 2))
print("2. 95% CI for Mean Glucose:",
      (round(lower_limit, 2), round(upper_limit, 2)))

print("3. Sample Proportion of Diabetes:", round(sample_proportion, 4))
print("4. 95% CI for Diabetes Proportion:",
      (round(lower_proportion, 4), round(upper_proportion, 4)))

print("5. t-statistic:", round(t_statistic, 4))
print("6. p-value:", round(p_value, 4))

if p_value < 0.05:
    print("7. Decision: Reject H0")
else:
    print("7. Decision: Fail to reject H0")
```
